```python
# Generate Telecom Network Inventory

device_types = [
    "Router",
    "Switch",
    "BTS",
    "Microwave",
    "Firewall",
    "OLT"
]

vendors = [
    "Cisco",
    "Nokia",
    "Ericsson",
    "Huawei",
    "Juniper"
]

regions = [
    "Bangalore",
    "Mumbai",
    "Delhi",
    "Chennai",
    "Hyderabad"
]

status = [
    "Healthy",
    "Warning",
    "Critical"
]

inventory = []

for i in range(1, 501):

    inventory.append({

        "Device_ID": f"DEV{i:04}",

        "Device_Type": random.choice(device_types),

        "Vendor": random.choice(vendors),

        "Region": random.choice(regions),

        "Status": random.choice(status)

    })

inventory_df = pd.DataFrame(inventory)

inventory_df.head()
```
